## creating a dataframe

```python
import pandas as pd
```

```python
data = {
    "name": ["rahul","anu","john"],
    "age":[20,21,19],
    "marks":[85,98,78]
}
df=pd.DataFrame(data)
print(df)
```

## accessing rows

```python
print(df.loc[0])
```

```python
print(df.iloc[1])
```

## functions

## sum

```python
(df["marks"].sum())
```

```python
(df["marks"].mean())
```

```python
(df["marks"].max())
```

## minimum

```python
(df["marks"].min())
```

## filtering data

```python
(df[df["marks"]>80])
```
