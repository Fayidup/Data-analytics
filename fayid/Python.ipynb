```python
score = int(input("enter the score:"))
if score >= 90 and score <=100:
    print("outstanding")
elif score >= 70 and score<=89:
    print("exceland")
elif score >= 60 and score<=69:
    print("good")
elif score >= 40 and score<=59:
    print("average")
elif score >= 20 and score<=39:
    print("pass")
elif score <=19 and score>=0:
    print("fail")
else :
    print("Invalid score")
    
```
