### Create and write to file

```python
with open("create.txt","w") as file:
    file.write("assalamu alaikum!")
```

### Read file

```python
with open("create.txt","r") as file:
    print(file.read())
```

### overwrite file

```python
with open("create.txt","w")as file:
    file.write("waalikumussalam!")
```

### append new line

```python
with open("create.txt","a")as file:
    file.write("\nNjammale peer muneer.")
```

### read full content again


```python
with open("create.txt","r")as file:
    print(file.read())
```

### read line by line

```python
with open("create.txt","r")as file:
    for line in file:
        print(line.strip())
```

```python
with open("create.txt","r")as file:
    print(file.readline())
```

### working with file pointer

```python
with open("create.txt","r") as file:
    print(file.read(6))
    file.seek(0)
    print(file.read())
```

```python
with open("create.txt","r")as file:
    print(file.tell())
    file .read(10)
    print(file.tell())
```

```python
with open("create.txt","r")as file:
    lines=file.readlines()
    print(lines)
    
```
